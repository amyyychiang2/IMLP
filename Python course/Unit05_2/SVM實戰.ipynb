* (一)引入函式庫及內建乳癌資料集<br>
引入之函式庫如下<br>
sklearn.datasets: 用來匯入內建之乳癌資料集`datasets.load_breast_cancer()`<br>
sklearn.SVR: 支持向量機回歸分析之演算法<br>
matplotlib.pyplot: 用來繪製影像

````python
from sklearn import svm
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
````

## Step1. 下載資料

````python
breast_cancer=datasets.load_breast_cancer()
````

````python
print(breast_cancer.keys())
print('Feature names:', breast_cancer.feature_names)
print('Target names:', breast_cancer.target_names)
print('Data shape:', breast_cancer.data.shape)
print('Target shape:', breast_cancer.target.shape)
````

````python
X = breast_cancer.data
y = breast_cancer.target

print('X shape:', X.shape)
print('y shape:', y.shape)
print('Target labels:', breast_cancer.target_names)
````

````python
import pandas as pd

df = pd.DataFrame(X, columns=breast_cancer.feature_names)
display(df.head())
print('Class counts (0=malignant, 1=benign):')
````

## Step2. 區分訓練集與測試集

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0, stratify=y #Preserve class proportions in both sets
)

print('Training set:', X_train.shape)
print('Test set:', X_test.shape)
````

## Step3. 建模

````python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler #standardize the data

clf = make_pipeline(
    StandardScaler(),
    svm.SVC(kernel='rbf', C=1.0, gamma='scale')
)
clf.fit(X_train, y_train)
````

## Step4. 預測

```

```


````python
y_pred = clf.predict(X_test)

print('First 20 predictions:', y_pred[:20])
print('First 20 actual labels:', y_test[:20])
````

## Step5. 準確度分析

````python
print('Training accuracy:', clf.score(X_train, y_train))
print('Test accuracy:', clf.score(X_test, y_test))
````

````python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

print('Accuracy:', accuracy_score(y_test, y_pred))
print('Confusion matrix:')
print(confusion_matrix(y_test, y_pred))
print('Classification report:')
print(classification_report(
    y_test, y_pred, target_names=breast_cancer.target_names
))
````

This shows that my model correctly classified 95.3% of the breast tumors in the test dataset as either malignant (cancerous) or benign (non-cancerous).
However, Of the 64 tumors that were actually malignant, the model incorrectly classified 4 as benign. In other words, when the model predicts cancer, it is correct approximately 94% of the time in this test set.
The recall rate shows that it detected approximately 94% of malignant tumors but missed approximately 6%.
F-1 score is 0.94 indicating a good balance between identifying malignant tumors and limiting false alarms.
