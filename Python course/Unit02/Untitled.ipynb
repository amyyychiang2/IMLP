```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
days=range(1,7)
celsius=[25.6,23.2,18.5,28.3,26.5,30.5]
plt.plot(days,celsius, "o-m")
plt.grid()
plt.title("daily temperature")
plt.show()
```

```python
x=np.arange(1,11)
y=[3,1,8,5,4,2,4,6,3,7]
sns.set(style="whitegrid")
plt.bar(x,y)
plt.show()
```
